  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432064_GalatiaSiregar
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Seed agar hasil dapat direproduksi
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

# Gunakan GPU jika tersedia
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', device)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432064_GalatiaSiregar
# 1. Muat dataset (file CSV dari Kaggle yang diunggah ke repositori GitHub / Colab)
DATA_PATH = 'finalgolddata.csv'   # sesuaikan dengan nama file CSV dataset
try:
    df = pd.read_csv(DATA_PATH)
except FileNotFoundError:
    from google.colab import files
    print('File tidak ditemukan, silakan unggah file CSV dataset:')
    uploaded = files.upload()
    df = pd.read_csv(list(uploaded.keys())[0])

print('Jumlah baris & kolom:', df.shape)
print('Nama kolom:', df.columns.tolist())
print(df.head())

# 2. Pembersihan data (format tanggal pada dataset: dd-mm-yy)
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%y', errors='coerce')
df = df.dropna(subset=['Date', 'Close'])
df = df.sort_values('Date').reset_index(drop=True)
print('Missing value pada kolom Date & Close setelah dibersihkan:', int(df[['Date', 'Close']].isnull().sum().sum()))
print('Rentang tanggal:', df['Date'].min().date(), 's/d', df['Date'].max().date())

dates = df['Date'].values
prices = df['Close'].values.astype(np.float32)

# 3. Pembagian data latih (80%) dan data uji (20%) secara berurutan (tanpa diacak)
WINDOW = 30
train_size = int(len(prices) * 0.8)

# 4. Sliding window + normalisasi relatif
#    Tiap jendela 30 hari dibagi dengan harga terakhir jendela tersebut, lalu dinyatakan dalam persen.
#    Cara ini dipilih karena harga emas pada data uji (hingga >5000 USD) jauh melampaui rentang data latih,
#    sehingga normalisasi Min-Max biasa akan membuat model gagal menjangkau harga tinggi.
def create_sequences(prices, window):
    X, y, base = [], [], []
    for i in range(len(prices) - window):
        win = prices[i:i + window]
        b = win[-1]                                   # harga hari terakhir pada jendela
        X.append((win / b - 1) * 100)                 # input: perubahan relatif (%) terhadap b
        y.append((prices[i + window] / b - 1) * 100)  # target: perubahan relatif (%) hari berikutnya
        base.append(b)                                # disimpan untuk mengembalikan ke harga asli
    return np.array(X)[..., None], np.array(y)[:, None], np.array(base)

X_all, y_all, base_all = create_sequences(prices, WINDOW)

# 5. Pisahkan berdasarkan hari yang diprediksi (target): index < train_size = latih, selebihnya = uji
split = train_size - WINDOW
X_train, y_train = X_all[:split], y_all[:split]
X_test, y_test, base_test = X_all[split:], y_all[split:], base_all[split:]

# 6. Konversi ke PyTorch Tensor
X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.float32)
X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.float32)

print('X_train:', tuple(X_train.shape), '| y_train:', tuple(y_train.shape))
print('X_test :', tuple(X_test.shape), '| y_test :', tuple(y_test.shape))

# 7. Visualisasi pembagian data latih dan data uji
plt.figure(figsize=(12, 4))
plt.plot(dates[:train_size], prices[:train_size], color='#114F6A', label='Data Latih')
plt.plot(dates[train_size:], prices[train_size:], color='#1FA4B1', label='Data Uji')
plt.title('Harga Penutupan Emas Harian (Close)')
plt.xlabel('Tanggal')
plt.ylabel('Harga (USD)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432064_GalatiaSiregar
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, num_layers=2, dropout=0.2, output_size=1):
        super(SequenceModel, self).__init__()
        # Layer LSTM: membaca urutan 30 hari dan menyimpan memori jangka panjang
        self.lstm = nn.LSTM(input_size=input_size, hidden_size=hidden_size,
                            num_layers=num_layers, batch_first=True,
                            dropout=dropout)
        # Layer Linear: mengubah hidden state terakhir menjadi 1 nilai prediksi
        self.fc = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        # x: (batch, window, 1) -> out: (batch, window, hidden_size)
        out, (h_n, c_n) = self.lstm(x)
        # Ambil output pada time step terakhir saja
        out = self.fc(out[:, -1, :])
        return out
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432064_GalatiaSiregar
# 1. Inisiasi hyperparameter
epochs = 50
learning_rate = 0.001
batch_size = 32
hidden_size = 64
num_layers = 2
dropout = 0.2

# 2. Inisiasi model, fungsi Loss, dan Optimizer
model = SequenceModel(input_size=1, hidden_size=hidden_size, num_layers=num_layers, dropout=dropout).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=batch_size, shuffle=True)

# 3. Riwayat nilai loss untuk divisualisasikan
train_losses = []
test_losses = []

# 4. Training loop
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()            # reset gradien
        pred = model(xb)                 # forward pass
        loss = criterion(pred, yb)       # hitung MSE
        loss.backward()                  # backpropagation
        optimizer.step()                 # update bobot
        running_loss += loss.item() * xb.size(0)
    epoch_train_loss = running_loss / len(train_loader.dataset)

    # Evaluasi pada data uji (tanpa update bobot)
    model.eval()
    with torch.no_grad():
        epoch_test_loss = criterion(model(X_test.to(device)), y_test.to(device)).item()

    train_losses.append(epoch_train_loss)
    test_losses.append(epoch_test_loss)
    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f'Epoch [{epoch + 1:>2}/{epochs}] | Train Loss: {epoch_train_loss:.6f} | Test Loss: {epoch_test_loss:.6f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432064_GalatiaSiregar
# 1. Plot grafik penurunan loss terhadap epoch
plt.figure(figsize=(10, 4))
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.plot(test_losses, color='#1FA4B1', linewidth=2, linestyle='--', label='Test Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE, satuan persen kuadrat)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 2. Prediksi pada data uji (perubahan relatif %), lalu kembalikan ke harga asli (USD)
model.eval()
with torch.no_grad():
    pred_pct = model(X_test.to(device)).cpu().numpy().flatten()
y_pred = base_test * (1 + pred_pct / 100)
y_true = prices[train_size:]

# 3. Metrik evaluasi
mae = np.mean(np.abs(y_true - y_pred))
rmse = np.sqrt(np.mean((y_true - y_pred) ** 2))
mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
r2 = 1 - np.sum((y_true - y_pred) ** 2) / np.sum((y_true - y_true.mean()) ** 2)
print(f'MAE  : {mae:.4f} USD')
print(f'RMSE : {rmse:.4f} USD')
print(f'MAPE : {mape:.2f} %')
print(f'R2   : {r2:.4f}')

# 4. Plot grafik perbandingan data aktual vs prediksi
test_dates = dates[train_size:]
plt.figure(figsize=(12, 5))
plt.plot(test_dates, y_true, color='#114F6A', linewidth=2, label='Data Aktual')
plt.plot(test_dates, y_pred, color='#1FA4B1', linewidth=2, label='Prediksi LSTM')
plt.title('Perbandingan Harga Emas Aktual vs Prediksi LSTM (Data Uji)')
plt.xlabel('Tanggal')
plt.ylabel('Harga Close (USD)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
#202432064_GalatiaSiregar
# 1. Baseline naive: prediksi harga besok = harga hari ini (base_test adalah harga terakhir tiap jendela)
y_naive = base_test

# 2. Hitung metrik baseline dengan rumus yang sama
mae_n = np.mean(np.abs(y_true - y_naive))
rmse_n = np.sqrt(np.mean((y_true - y_naive) ** 2))
mape_n = np.mean(np.abs((y_true - y_naive) / y_true)) * 100
r2_n = 1 - np.sum((y_true - y_naive) ** 2) / np.sum((y_true - y_true.mean()) ** 2)

# 3. Tabel perbandingan LSTM vs baseline
hasil = pd.DataFrame({
    'Model': ['LSTM', 'Baseline Naive'],
    'MAE (USD)': [mae, mae_n],
    'RMSE (USD)': [rmse, rmse_n],
    'MAPE (%)': [mape, mape_n],
    'R2': [r2, r2_n]
}).round(4)
print(hasil.to_string(index=False))

# 4. Grafik zoom: 100 hari terakhir data uji
n_zoom = 100
plt.figure(figsize=(12, 5))
plt.plot(test_dates[-n_zoom:], y_true[-n_zoom:], color='#114F6A', linewidth=2, marker='o', markersize=3, label='Data Aktual')
plt.plot(test_dates[-n_zoom:], y_pred[-n_zoom:], color='#1FA4B1', linewidth=2, marker='x', markersize=4, label='Prediksi LSTM')
plt.title('Zoom 100 Hari Terakhir: Aktual vs Prediksi LSTM')
plt.xlabel('Tanggal')
plt.ylabel('Harga Close (USD)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
